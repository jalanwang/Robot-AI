```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

x=np.linspace(0, 2*np.pi,1000)
y1=np.sin(x)
y2=np.cos(x)

plt.title('Sin cos curve')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.plot(x,y1,'b', label='sin curve')
plt.plot(x,y2,'r', label='cos curve' )

plt.legend()
plt.savefig("sin_cos.png")
```

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2) # 2행 2열의 그림을 그리는 공간을 만든다

X = np.random.randn(100)     # 정규 분포를 가지는 데이터
Y = np.random.randn(100)     # 정규 분포를 가지는 데이터
ax[0, 0].scatter(X, Y)       # 산점도 그림
X = np.arange(10)            # 0에서 9 사이의 연속값
Y = np.random.uniform(1, 10, 10) # 균일분포값 생성
ax[0, 1].bar(X, Y)           # 막대 차트 그림
X = np.linspace(0, 10, 100)
Y = np.cos(X)
ax[1, 0].plot(X, Y)          # 실선으로 함수를 그림|
Z = np.random.uniform(0, 1, (5, 5))
ax[1, 1].imshow(Z)           # 분포를 2D 이미지로 그림
```

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

f1 = np.random.randn(100000)
plt.hist(f1, bins=200, color='red', alpha=.7,\
         label='avg = 0, std = 1') # 표준편차가 standard deviation=1
plt.axis([-8, 8, 0, 2500])
plt.legend()
```

```python

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

f1 = np.random.normal(loc=0, scale=1, size=100000)
f2 = np.random.normal(loc=3, scale=1, size=100000)

plt.hist(f1, bins=200, color='red', alpha=.8,\
         label='avg = 0, std = 1')
plt.hist(f2, bins=200, color='green', alpha=.8,\
         label='avg = 3, std = 1')
plt.axis([-8, 8, -2, 2500])
plt.legend()
```
