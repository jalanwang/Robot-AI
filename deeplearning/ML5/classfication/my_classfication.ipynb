### AND와 OR 회로를 퍼셉트론으로 만들자

```python
import numpy as np

def activation(s):
   if s > 0:
       return 1
   else :
       return -1

def perceptron(W, x, b):
   # W와 x의 내적을 계산하고 편향 b를 더함
   tmp = np.dot(W, x) + b
   return activation(tmp)
```

```python
print('--- 퍼셉트론으로 구현한 AND 회로 ---')
W = np.array([0.5, 0.5])
b = -0.7
for x1, x2 in [(-1, -1), (-1, 1), (1, -1), (1, 1)]:
    x = np.array([x1, x2])
    y = perceptron(W, x, b)
    print('{:3d} {:3d} : {:3d}'.format(x1, x2, y))
```

```python
import numpy as np

def activation(s):
   if s > 0:
       return 1
   else :
       return 0

def perceptron(x1, x2, b=-0.7):
   W = np.array([x1, x2])
   tmp = np.sum( W * x ) + b # np.dot(W,x)+b
   #print(x1,'*',W[0],'+',x2,'*',W[1],'+', b, '=', tmp)
   return activation(tmp)

   print('--- 퍼셉트론으로 구현한 AND 회로 ---')
W = np.array([0.5, 0.5])
b = -0.7
for x1, x2 in [(0, 0), (0, 1), (1, 0), (1, 1)]:
    y = perceptron(x1, x2)
    print('{:3d} {:3d} : {:3d}'.format(x1, x2, y))
```

```python
from sklearn.neural_network import MLPClassifier

from sklearn.linear_model import Perceptron

X=[[0,0], [0,1], [1,0], [1,1]] #입력 값
Y=[0,0,0,1] #AND 게이트의 출력 값
perception = Perceptron()
perception.fit(X, Y)
print(perception.predict(X))

```

```python
print(perception.coef_)
print(perception.intercept_)

perception.predict([[0,0], [0,1], [1,0], [1,1]])
```

```python
X1=[[0,0], [0,1], [1,0], [1,1]] #입력 값
Y1=[0,1,1,1] #OR 게이트의 출력 값
perception1 = Perceptron()
perception1.fit(X1, Y1)

print(perception1.coef_)
print(perception1.intercept_)

print(perception1.predict(X1))
```
