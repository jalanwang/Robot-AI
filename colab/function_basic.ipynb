# Python 매개변수 종류

## 1. 기본매개변수 (Default Parameters)
## 2. 가변매개변수 (Variable-length Parameters) - *args
## 3. 키워드매개변수 (Keyword Parameters) - **kwargs

## 1. 기본매개변수 (Default Parameters)
함수를 호출할 때 인자를 생략하면 기본값이 사용됩니다.

```python
# 기본매개변수 예시
def greet(name, message="안녕하세요"):
    """
    name: 필수 매개변수
    message: 기본값이 "안녕하세요"인 기본매개변수
    """
    print(f"{message}, {name}님!")

# 사용 예시
greet("철수")  # message는 기본값 사용
greet("영희", "반갑습니다")  # message를 직접 지정
```

## 2. 가변매개변수 (Variable-length Parameters) - *args
개수가 정해지지 않은 여러 개의 인자를 받을 수 있습니다. 튜플 형태로 전달됩니다.

```python
# 가변매개변수 예시
def sum_all(*numbers):
    """
    *numbers: 여러 개의 숫자를 받을 수 있는 가변매개변수
    """
    print(f"받은 인자들: {numbers}")  # 튜플 형태
    print(f"타입: {type(numbers)}")
    return sum(numbers)

# 사용 예시
print(f"결과: {sum_all(1, 2, 3)}")
print(f"결과: {sum_all(10, 20, 30, 40, 50)}")
print(f"결과: {sum_all(100)}")
```

## 3. 키워드매개변수 (Keyword Parameters) - **kwargs
key=value 형태로 인자를 받을 수 있습니다. 딕셔너리 형태로 전달됩니다.

```python
# 키워드매개변수 예시
def print_info(**kwargs):
    """
    **kwargs: key=value 형태의 여러 인자를 받을 수 있는 키워드매개변수
    """
    print(f"받은 인자들: {kwargs}")  # 딕셔너리 형태
    print(f"타입: {type(kwargs)}")
    print("\n정보 출력:")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

# 사용 예시
print_info(이름="김철수", 나이=25, 직업="개발자")
print("\n")
print_info(도시="서울", 국가="대한민국", 인구=1000)
```

## 모두 함께 사용하기
세 가지 매개변수를 모두 함께 사용할 수 있습니다.
순서: 일반매개변수 → 기본매개변수 → *args → **kwargs

```python
# 모든 매개변수 타입을 함께 사용
def order_food(customer, store="맥도날드", *items, **options):
    """
    customer: 일반 매개변수 (필수)
    store: 기본매개변수 (기본값 있음)
    *items: 가변매개변수 (여러 메뉴)
    **options: 키워드매개변수 (추가 옵션)
    """
    print(f"=== 주문 정보 ===")
    print(f"고객: {customer}")
    print(f"매장: {store}")
    print(f"주문 메뉴: {', '.join(items)}")
    print(f"추가 옵션:")
    for key, value in options.items():
        print(f"  - {key}: {value}")

# 사용 예시
order_food("홍길동", "버거킹", "와퍼", "콜라", "감자튀김",
           배달="예", 결제="카드", 요청사항="빨대 넣어주세요")
```

**다음 빈칸을 채워 매개변수로 전달된 값들을 모두 곱해서 리턴하는 가변 매개변수 함수를 만들어 보세요.**

```python
def mul(*values):
  ret=1
  for value in values:
    ret=ret*value
  return(ret)

print(mul(5,7,9,10))
```

**재귀호출**
아랫처럼 해야 한 번 더 도는 루프를 방지할 수 있다.
if n==0으로 하면 한 바퀴 불필요하게 더 돈다. n==0이면 n!이 1이라는 수학적 정의가 아니므로 n==1로 끝내는게 바람직하다고 본다.

```python
def factorial(n):
  if n==1:
    return 1
  else:
    return n*factorial(n-1)

print(factorial(5))
```
