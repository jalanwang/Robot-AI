**윗 코드 앞뒤로 인사의 시작 및 종료를 나타내려면?**

**decorator 연습**



```python
def test(func):
    print('인사의 시작')
    func()
    print('인사의 종료')

def hello():
    print('hello by normal function call')

test(hello)
```

```python
def test(func):
    def wrapper():
        print('인사가 시작되었습니다.')
        func()
        print('인사가 종료되었습니다')
    return wrapper

@test
def hello():
    print('hello by decorator')

hello()
```
