```python
import random

print(random.random())

print(round(random.uniform(10,20),2))

print(random.randrange(10))

print(random.choice([1,2,3]))

lotto_pool= [1,2,3,4,5]
print(random.shuffle(lotto_pool)) #반환값이 없는 메소드임
print(lotto_pool)

print(random.sample([1,2,3,4,5],5))

```
