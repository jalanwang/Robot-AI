### format02.py

```python
output_a="{:d}".format(42)

output_b="{:05d}".format(42)
output_c="{:10d}".format(42)

print(output_a)  # 출력: 42
print(output_b)  # 출력: 00042
print(output_c)  # 출력:         42

output_d="{:.2f}".format(3.14159)
output_e="{:.4f}".format(3.14159)

print(output_d)  # 출력: 3.14
print(output_e)  # 출력: 3.1416

output_f="{:+015f}".format(3.14159)
print(output_f)  # 출력: +0000003.141590
```
