```python
# tensorflow와 tf.keras를 임포트
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt

shape=(4,28,28,3)
X=tf.random.normal(shape)
y=tf.keras.layers.Conv2D(2,3,activation='relu',input_shape=(28,28,3))(X)
# Conv2D의 첫번째 인자는 뉴런의 갯수, 두번째 인자는 커널의 크기, activation은 활성화 함수, input_shape는 입력 데이터의 형태
# 윗 2이라는 값이 뉴런의 갯수
print(y.shape)
```

```python
import matplotlib.pyplot as plt
import tensorflow as tf
(train_images, train_labels),(test_images, test_labels) = tf.keras.datasets.mnist.load_data()
```

```python
train_images.reshape(-1,28,28,1)
test_images.reshape(-1,28,28,1)
```

```python
(train_images, test_images) = train_images / 255.0, test_images / 255.0
```

```python
model = keras.Model.Sequential()

model.add(layers.Input((28,28,1)))
model.add(layers.Conv2D(32, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))

model.add(layers.Flatten()) # fully connected layer로 바꿔주는 역할
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.fit(train_images, train_labels, epochs=5)
```
