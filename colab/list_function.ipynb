```python

numbers=[1,2,3,4,5]

min(numbers), max(numbers), sum(numbers)
```

```python
reversed(numbers)
```

```python
list_reversed = reversed(numbers)

print('# reversed() 함수 결과')
print(list_reversed)

print('list(reversed()) 함수 결과')
print(list(list_reversed))
```
