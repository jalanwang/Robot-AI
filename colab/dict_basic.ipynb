```python
dictionary = {
  "fruit": "apple",
  "color": "red",
  "quantity": 5
}

key=input("Enter key to lookup (fruit, color, quantity): ")
if key in dictionary:
  print("Value:", dictionary[key])
else:
  #print("Key not found in the dictionary.")
  raise KeyError("Key not found in the dictionary.")
```
